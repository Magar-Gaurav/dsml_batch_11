# **Evaluation Metrics**

**Condition: Loan Predict System**
- what accuracy do you get from your system?

**This type of problem is called Classification problem**
  
  *Accuracy if basically the number of correct prediction out of total prediction*
  
  

  **Formula for finding accuracy**
  - accuracy = (Correct prediction) / (Total Prediction)

  *precision is basically the number of true predicted by sum of true prediction and false positive*

  **Formula for finding percision**
  - precision = (Correct prediction) / (correct prediction + false positive prediction)


  *recall is basically true positive prediction by True Positive and False Negative*

  **Formula for finding recall**
  
  - recall = (Correct Prediction) / (True Positive  + False Negative)

  *F1 score is harmonic mean of precision and recall*
  
  **Formula for finding f1 score**
  
  - f1_score = 2 * (precision * recall) / (precision + recall)

**function for calculating performance metrics**

```python
def find_accuracy(t_p, t_n, f_p, f_n):
    return (t_p + t_n) / (t_p + t_n + f_p + f_n)
```

```python
def find_precision(t_p, f_p):
    return t_p / (t_p + f_p)
```

```python
def find_recall(t_p, f_n):
    return t_p / (t_p + f_n)
```

```python
def find_f1_score(t_p, f_p, f_n):
    precision = find_precision(t_p, f_p)
    recall = find_recall(t_p, f_n)
    return 2 * (precision * recall) / (precision + recall)
```

```python
print("Accuracy: ",find_accuracy(50,25,15,10))
print("Precision: ",find_precision(50,15))
print("Recall: ",find_recall(50,10))
print("F1 Score: ",find_f1_score(50,10,15))
```
