```python
# pip install -U spacy
# python -m spacy download en_core_web_sm
import spacy

# Load English tokenizer, tagger, parser and NER
nlp = spacy.load("en_core_web_sm")

# Process whole documents
text = ("When Sebastian Thrun started working on self-driving cars at "
        "Google in 2007, few people outside of the company took him "
        "seriously. “I can tell you very senior CEOs of major American "
        "car companies would shake my hand and turn away because I wasn’t "
        "worth talking to,” said Thrun, in an interview with Recode earlier "
        "this week.")
doc = nlp(text)

# Analyze syntax
print("Noun phrases:", [chunk.text for chunk in doc.noun_chunks])
print("Verbs:", [token.lemma_ for token in doc if token.pos_ == "VERB"])

# Find named entities, phrases and concepts
for entity in doc.ents:
    print(entity.text, entity.label_)
```

```python
doc = nlp("Apple is looking at buying a startup.")
```

```python
for token in doc:
  print(token.text, token.dep_, token.head.text)
```

# **Semantic level**

image.png

```python
import nltk
from nltk.wsd import lesk
from nltk.corpus import wordnet

# Make sure required resources are downloaded
nltk.download("wordnet")
nltk.download("omw-1.4")

# Example 1: financial sense of "bank"
sentence1 = "I deposited money in the bank".split()
sense1 = lesk(sentence1, "bank")
print("Sentence 1 sense:", sense1)
print("Definition:", sense1.definition())

# Example 2: river-side sense of "bank"
sentence2 = "Tall green grasses are on the bank of the river".split()
sense2 = lesk(sentence2, "bank")
print("\nSentence 2 sense:", sense2)
print("Definition:", sense2.definition())
```

# **Discourse level**

image.png

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "John brought a laptop. He loves it."
doc = nlp(text)

for token in doc:
    print(token.text, token.pos_, token.dep_, token.head.text)
```
