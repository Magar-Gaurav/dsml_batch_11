**downloading NLTK data**

```python
import nltk
nltk.download('punkt') # download the tokenizer model
nltk.download('punkt_tab')
```

**Breaking words into token**

```python
from nltk.tokenize import word_tokenize
text = "I love learning AI."
tokens = nltk.word_tokenize(text)
print(f"Tokens:\n{tokens}")
```

**POS (Part of Speech)**

```python
nltk.download("averaged_perceptron_tagger_eng")

text = "John is running fast."

tokens = nltk.word_tokenize(text)
pos = nltk.pos_tag(tokens)
print(pos)
```

**Stop words**

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')


text = "I am from DSML class"
tokens = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in tokens if word.lower() not in stop_words]

print("Original Text: ", text)
print("Tokens: ", tokens)
print("Stop words: ", stop_words)
print("Filtered words (without stopwords): ",filtered_words)
```

**Word Frequency Analysis**

```python
from collections import Counter

text = "apple apple banana mango apple banana"

words = text.split(" ")
word_freq = Counter(words)
print("Word frequency:\n",word_freq)
```

**NER (Named Entity Recognition)**

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Apple is looking at buying a U.K. startup for $1 billion in 2026."
doc = nlp(text)

for ent in doc.ents:
  print(f"{ent.label_} =  {ent.text}")
```

**Sentimental Analysis**

```python
from textblob import TextBlob

text = "I love this phone."

blob = TextBlob(text)

print(blob.sentiment)

# polarity > 0 = positive
# polarity < 0 = negative
# polarity = 0 = neutral
```

**TF-IDF (Term Frequency - Inverse Document Frequency)**

image.png

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love AI",
    "AI is amazing",
    "I love programming"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

**Bag of words**

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love AI",
    "AI is amazing"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```
