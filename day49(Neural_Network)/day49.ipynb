```python
import torch
import torch.nn as nn
```

```python
import torch.optim as optim

# Define the Feed Forward Neural Network
class FeedForwardNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(FeedForwardNN, self).__init__()

        # Define layers
        self.fc1 = nn.Linear(input_size, hidden_size)  # First hidden layer
        self.relu = nn.ReLU()                          # Activation function
        self.fc2 = nn.Linear(hidden_size, output_size) # Output layer

    def forward(self, x):
        out = self.fc1(x)   # Input -> Hidden
        out = self.relu(out) # Apply activation
        out = self.fc2(out) # Hidden -> Output
        return out

# Example usage
if __name__ == "__main__":
    # Hyperparameters
    input_size = 10
    hidden_size = 20
    output_size = 3
    learning_rate = 0.001

    # Create model
    model = FeedForwardNN(input_size, hidden_size, output_size)
    print(model)

    # Example data (batch_size=5, input_size=10)
    x = torch.randn(5, input_size)
    y = torch.randint(0, output_size, (5,))  # Random class labels

    # Loss and optimizer
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)

    # Forward pass
    outputs = model(x)
    loss = criterion(outputs, y)

    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    print(f"Loss: {loss.item():.4f}")
```
