**Concatenation**

```python
df1 = pd.DataFrame({
    "acc" : [1,2,3],
    "name" : ["Saurav","Gaurav","Sandesh"],
    "age" : [24,23,19]
})

df2 = pd.DataFrame({
    "acc" : [4,5],
    "name" : ["Gyanu","Daya"],
    "age" : [43,50]
})
```

```python
row_df = pd.concat([df1,df2])
row_df = row_df.set_index("acc")
row_df
```

```python
df3 = pd.DataFrame({
    "school" : ["JMC","KMC","Saraswati"]
})
```

```python
column_df = pd.concat([df1,df3], axis = 1)
column_df
```

**Merge**

```python
new_df1 = pd.DataFrame({
    "key" : ["E","B","C","Z"],
    "inv_values" : [99,55,33,22],
    "account" : [1,2,5,7]
})
```

```python
new_df2 = pd.DataFrame({
    "key" : ["A","B","C","D"],
    "hel_ration" : [66,0.67,67,20],
    "account" : [11,32,53,74]
})
```

```python
merged_df = pd.merge(new_df1,new_df2,how="inner",on="key",suffixes=("df1","df2"))
merged_df
```

**Class task: doing all types of joins**

```python
outer_merged_df = pd.merge(new_df1,new_df2,how="outer",on="key",suffixes=("df1","df2"))
outer_merged_df
```

```python
right_merged_df = pd.merge(new_df1,new_df2,how="right",on="key",suffixes=("df1","df2"))
right_merged_df
```

```python
left_merged_df = pd.merge(new_df1,new_df2,how="left",on="key",suffixes=("df1","df2"))
left_merged_df
```

```python
cross_merged_df = pd.merge(new_df1,new_df2,how="cross")
cross_merged_df
```

**join**

```python
new_df1.join(new_df2,how="inner",lsuffix="_x")
```

```python
new_df2.join(new_df1,how = "right",lsuffix = "_x")
```

**Stacking**

```python
new_df1 = pd.DataFrame({
    "key" : ["E","B","C","Z"],
    "inv_values" : [99,55,33,22],
    "account" : [1,2,5,7]
}, index = ["W","X","Y","Z"]
)

new_df1
```

```python
stacked = new_df1.stack()
stacked
```

```python
unstacked = stacked.unstack()
unstacked
```
