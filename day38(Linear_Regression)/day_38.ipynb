**converting into numpy**

```python
X = data_df.TV.to_numpy()
y = data_df.sales.to_numpy()
```

```python
x_mean = X.mean()
y_mean = y.mean()
```

```python
beta_1 = np.sum((X - x_mean) * (y - y_mean)) / np.sum(np.power((X-x_mean),2))
beta_0 = y_mean - (beta_1 * x_mean)
```

```python
print(f"y = {beta_0:.2f} + {beta_1:.2f}x")
```

```python
plt.scatter(X,y,color = "green")
plt.plot(X,beta_0 + (beta_1 * X),color = "black")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def train_plot(data_df, feature, ax, c):
    # Extract arrays
    X = data_df[feature].to_numpy()
    Y = data_df["sales"].to_numpy()

    # Means
    x_mean = X.mean()
    y_mean = Y.mean()

    # Coefficients
    beta_1 = np.sum((X - x_mean) * (Y - y_mean)) / np.sum((X - x_mean)**2)
    beta_0 = y_mean - (beta_1 * x_mean)

    # Regression line
    y_hat = beta_0 + beta_1 * X

    # Plot
    ax.scatter(X, Y, color=c, marker=".")
    ax.plot(X, y_hat, color="black")
    ax.set_xlabel(feature)
    ax.set_ylabel("Sales")
    ax.set_title(f"$y$ = {beta_0:.2f} + {beta_1:.2f}$x$")

# Example: plotting all three predictors in your GridSpec layout
fig = plt.figure(figsize=(15,4))
gs = plt.GridSpec(1,3)

ax0 = fig.add_subplot(gs[0,0])
train_plot(data_df, "TV", ax0, "green")

ax1 = fig.add_subplot(gs[0,1])
train_plot(data_df, "radio", ax1, "blue")

ax2 = fig.add_subplot(gs[0,2])
train_plot(data_df, "newspaper", ax2, "orange")

plt.tight_layout()
plt.show()
```
